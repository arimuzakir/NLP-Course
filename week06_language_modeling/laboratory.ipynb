# Week 6 - Bigram Language Model

```python
from collections import Counter
import math
corpus=['saya belajar nlp','saya belajar machine learning','mahasiswa belajar nlp','dosen mengajar nlp']
uni,bi=Counter(),Counter()
for s in corpus:
    t=['<s>']+s.split()+['</s>']
    for x in t: uni[x]+=1
    for a,b in zip(t,t[1:]): bi[(a,b)]+=1

def score(s,alpha=1):
    t=['<s>']+s.split()+['</s>']; V=len(uni); total=0
    for a,b in zip(t,t[1:]): total+=math.log((bi[(a,b)]+alpha)/(uni[a]+alpha*V))
    return total
for s in ['saya belajar nlp','mahasiswa belajar nlp','dosen belajar nlp']: print(s,score(s))
```
