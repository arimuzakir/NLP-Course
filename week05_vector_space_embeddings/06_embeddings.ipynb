# Word Embeddings

```python
from sklearn.feature_extraction.text import CountVectorizer
corpus=['kucing makan ikan','kucing tidur di rumah','anjing makan daging','anjing tidur di rumah','mahasiswa membaca buku','mahasiswa belajar NLP']
X=CountVectorizer().fit_transform(corpus).toarray(); print(X.shape)
```

## Latihan

Jelaskan mengapa count vector berbeda dari dense embedding seperti Word2Vec. Jika menggunakan pretrained embedding, periksa dimensi, OOV, similarity, dan potensi bias domain.
