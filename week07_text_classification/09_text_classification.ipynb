# Text Classification Baseline

```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report
root=Path.cwd().parents[0]; data_dir=root/'data'/'text_classification'/'bbc'; rows=[]
for cat in sorted(p for p in data_dir.iterdir() if p.is_dir()):
 for f in cat.glob('*.txt'): rows.append({'text':f.read_text(encoding='utf-8',errors='ignore'),'label':cat.name})
df=pd.DataFrame(rows); Xtr,Xte,ytr,yte=train_test_split(df.text,df.label,test_size=.2,random_state=42,stratify=df.label); print(df.label.value_counts())
```

```python
models={'Logistic Regression':LogisticRegression(max_iter=1000),'Naive Bayes':MultinomialNB(),'Linear SVM':LinearSVC()}
for name,m in models.items():
 p=Pipeline([('tfidf',TfidfVectorizer(ngram_range=(1,2),min_df=2)),('model',m)]); p.fit(Xtr,ytr); pred=p.predict(Xte); print('\n###',name); print(classification_report(yte,pred))
```
