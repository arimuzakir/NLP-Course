# Semantic Similarity Baseline

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
pairs=[('Mahasiswa membaca buku NLP.','Mahasiswa sedang mempelajari NLP dari buku.'),('Model klasifikasi mendeteksi spam.','Sistem menemukan pesan yang tidak diinginkan.'),('Kucing duduk di kursi.','Harga saham meningkat hari ini.')]
texts=[x for p in pairs for x in p]; X=TfidfVectorizer().fit_transform(texts)
for i,(a,b) in enumerate(pairs): print(i+1, round(float(cosine_similarity(X[2*i],X[2*i+1])[0,0]),4), a,'||',b)
```

Jika dua kalimat bermakna mirip tetapi tidak berbagi banyak kata, mengapa TF-IDF dapat gagal? Hubungkan dengan distributional semantics.
