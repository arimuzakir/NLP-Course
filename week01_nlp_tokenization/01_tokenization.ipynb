# Tujuan

Memahami unit analisis NLP dan konsekuensi pemilihan tokenizer.

```python
import re
text="NLP memproses bahasa manusia: misalnya, 'AI-based systems' dan kata-kata gaul."
def whitespace_tokenize(t): return t.split()
def regex_tokenize(t): return re.findall(r'\w+|[^\w\s]',t,flags=re.UNICODE)
print(whitespace_tokenize(text))
print(regex_tokenize(text))
```

## Latihan

Tambahkan contoh URL, hashtag, emotikon, angka, singkatan, dan kata berimbuhan. Bandingkan hasil dan jelaskan tokenizer yang paling sesuai untuk tiap kasus.
