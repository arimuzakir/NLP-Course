# Week 1 - Tokenization Laboratory

```python
import re, pandas as pd
samples=['Mahasiswa belajar NLP.','@kampus #NLP keren banget!','Email dosen@example.com','harga 10,5%','kata gaul: nggak bangettt :)']
pd.DataFrame({'text':samples})
```

```python
def whitespace_tokenize(t): return t.split()
def regex_tokenize(t): return re.findall(r'\w+|[^\w\s]',t,flags=re.UNICODE)
for s in samples:
    print('\n',s); print('whitespace:',whitespace_tokenize(s)); print('regex:',regex_tokenize(s))
```
