# Dataset BBC

```python
from pathlib import Path
import pandas as pd
root=Path.cwd().parents[0]; data_dir=root/'data'/'text_classification'/'bbc'
rows=[]
for cat in sorted(p for p in data_dir.iterdir() if p.is_dir()):
    for f in sorted(cat.glob('*.txt'))[:30]: rows.append({'text':f.read_text(encoding='utf-8',errors='ignore'),'label':cat.name})
df=pd.DataFrame(rows); print(df.label.value_counts())
```

```python
import re
def conservative(t): return re.sub(r'\s+',' ',t.replace('\n',' ')).strip()
def aggressive(t):
    t=conservative(t).lower(); t=re.sub(r'http\S+|www\.\S+',' ',t); t=re.sub(r'[^a-z0-9\s]',' ',t); return re.sub(r'\s+',' ',t).strip()
s=df.iloc[0].text
print(s[:300]); print('\n',conservative(s)[:300]); print('\n',aggressive(s)[:300])
```
